The purpose of this notebook is to create a csv that has the cleaned values for the target variables. The previous dataset contained rows which had the total of all rows at the end, which greatly skewed the average counts and also created outliers.

```python
import numpy as np
import pandas as pd
```

```python
from google.colab import files
import io
import pandas as pd
from pathlib import Path

#Uploading the xlsx file
uploaded = files.upload()
```

```python
# Reading the sheets from the uploaded file, using excel_bytes
excel_bytes = uploaded['2007-2023_PIT_Data (1).xlsx']
sheets = pd.read_excel(io.BytesIO(excel_bytes), sheet_name=None)

# I used the data transform tool in Excel to get the columns that I wanted from the 2007 sheet, this way I can just use the columns in the 2007 sheet as reference for other sheets
ref_cols = list(sheets['2007'].columns)

# Filtering the sheets, and adding a year column which is from the name of the sheet
df_list = []
for year, df in sheets.items():
    keep = [c for c in ref_cols if c in df.columns]
    df = df[keep]
    df['Year'] = year
    df_list.append(df)

# Concat into one cleaned dataset, dropna for hanging rows from the original dataset.
combined = pd.concat(df_list, ignore_index=True)
combined_clean = combined.dropna()


# Saving the combined csv file
out_csv = "combined_with_year.csv"
combined_clean.to_csv(out_csv, index=False)
files.download(out_csv)
```

```python
# Checking the csv file now to make sure that the max values make sense
numeric_cols = combined_clean.select_dtypes(include='number').columns

# 2. Compute max values per numeric column
max_values = combined_clean[numeric_cols].max()
print("Max values per numeric column:")
print(max_values)
```
