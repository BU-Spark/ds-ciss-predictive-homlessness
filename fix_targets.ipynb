```python
base_path = '/content/drive/My Drive/DS539'

df_path = f'{base_path}/CISS_Final_Dataset.csv'
```

```python
import pandas as pd

final_df = pd.read_csv(df_path)
```

```python
final_df.head()
```

```python
final_df = final_df.drop(columns=['tract count'])

cols_to_modify = [
    'Sheltered Total Homeless',
    'Unsheltered Homeless',
    'Overall Homeless Individuals',
    'Overall Homeless People in Families',
    'Overall Homeless'
]

for col in cols_to_modify:
    final_df[col] = (final_df[col] / final_df['total population']) * 1000
```

```python
final_df.to_csv('CISS_final_df2.csv', index=False)
```
