- With dataset 3, remove the acs(rent, income, population) data and PIT data(homeless targets).
- Merge by year and CoC Number

```python
import pandas as pd
```

```python
df1 = pd.read_csv('/content/corrected_stuff.csv')
df2 = pd.read_csv('/content/combined_with_year (1).csv')
df3 = pd.read_csv('/content/Final_Dataset 3.csv')
```

```python
df1.rename(columns = {'COCNUM' : 'CoC_Number', 'year':'Year'}, inplace = True)
```

```python
df2.rename(columns = {'CoC Number': 'CoC_Number'}, inplace = True)
```

```python
df3 = df3.drop(columns = ["Total Population", "Median Gross Rent", "Median Household Income", "Overall Homeless",
                "Overall Homeless Individuals", "Overall Homeless People in Families", "Unsheltered Homeless", "Sheltered Total Homeless",
                          "Overall_Homeless_Per_Capita", "Overall_Homeless_Individuals_Per_Capita", "Overall_Homeless_People_in_Families_Per_Capita",
                          "Unsheltered_Homeless_Per_Capita", "Sheltered_Homeless_Per_Capita", "Total Year-Round Beds (ES, TH, SH)"])

df3.head()
```

```python
df3.rename(columns = {'CoC Number': 'CoC_Number'}, inplace = True)
df3.head()
```

```python
print(df1.info())
print(df2.info())
print(df3.info())
```

```python
# do an outer merge with an “_merge” flag
diff = df1.merge(
    df2,
    on=['CoC_Number', 'Year'],
    how='outer',
    indicator=True
)

# all pairs only in df1 (dropped from the inner join)
dropped_from_df2 = diff[diff['_merge'] == 'left_only'][['CoC_Number','Year']]
print(f"Lost from df2 ({len(dropped_from_df2)} pairs):")
print(dropped_from_df2)
```

```python
pairs = list(zip(dropped_from_df2['CoC_Number'], dropped_from_df2['Year']))
```

```python
print(pairs)
print(len(pairs))
```

```python
Cdf12 = pd.merge(df1, df2, on=['CoC_Number', 'Year'], how = 'inner')
print(df12.head())
print(df12.info())
```

```python
df_all = pd.merge(df12,df3, on= ['CoC_Number', 'Year'], how = 'left')
print(df_all.info())
df_all.head()
```

```python
df_all.to_csv('CISS_Final_Dataset', index = False)
```
